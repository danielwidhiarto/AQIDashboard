```python
import pandas as pd
import matplotlib.pyplot as plt

# Baca data dari CSV
df_wanliu = pd.read_csv("data/wanliu.csv")
df_changping = pd.read_csv("data/changping.csv")

# Hitung rata-rata PM2.5 per bulan untuk setiap lokasi
wanliu_pm25_monthly_avg = df_wanliu.groupby('month')['PM2.5'].mean()
changping_pm25_monthly_avg = df_changping.groupby('month')['PM2.5'].mean()

# Plot perbandingan rata-rata PM2.5
plt.figure(figsize=(10, 6))
wanliu_pm25_monthly_avg.plot(label='Wanliu')
changping_pm25_monthly_avg.plot(label='Changping')
plt.xlabel('Month')
plt.ylabel('Average PM2.5 Level')
plt.title('Comparison of Monthly Average PM2.5 Level between Wanliu and Changping')
plt.legend()
plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```
