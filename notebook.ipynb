```python
import pandas as pd
import matplotlib.pyplot as plt

# Baca data dari CSV
df_wanliu = pd.read_csv("data/wanliu.csv")
df_changping = pd.read_csv("data/changping.csv")

# Hitung rata-rata PM2.5 per bulan untuk setiap lokasi
wanliu_pm25_monthly_avg = df_wanliu.groupby('month')['PM2.5'].mean()
changping_pm25_monthly_avg = df_changping.groupby('month')['PM2.5'].mean()

# Plot perbandingan rata-rata PM2.5
plt.figure(figsize=(10, 6))
wanliu_pm25_monthly_avg.plot(label='Wanliu')
changping_pm25_monthly_avg.plot(label='Changping')
plt.xlabel('Month')
plt.ylabel('Average PM2.5 Level')
plt.title('Comparison of Monthly Average PM2.5 Level between Wanliu and Changping')
plt.legend()
plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

```python
# Buat subplots untuk setiap jam
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 10))
fig.suptitle('Comparison of NO2 Levels between Wanliu and Changping by Hour', fontsize=16)

# Loop melalui setiap jam
for i, hour in enumerate(range(9)):
    ax = axes[i//3, i%3]
    wanliu_no2_hourly = df_wanliu[df_wanliu['hour'] == hour]['NO2']
    changping_no2_hourly = df_changping[df_changping['hour'] == hour]['NO2']
    ax.hist([wanliu_no2_hourly, changping_no2_hourly], bins=15, alpha=0.5, label=['Wanliu', 'Changping'])
    ax.set_title(f'Hour {hour}')
    ax.legend()

plt.tight_layout()
plt.show()
```
